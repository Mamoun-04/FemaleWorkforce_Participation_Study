# Education, Workforce, and Policies: Drivers of Economic and Social Change

# Abstract

In this study, we aimed to explore the interplay between education systems, female workforce participation, and government policies, and their impact on economic growth and social stability across eight nations: Japan, China, Germany, the US, the UK, France, Saudi Arabia, and Brazil. By examining variables such as education enrollment rates, female labor force participation, GDP growth, and government spending on education, we sought to identify key factors that contribute to national progress.

Our analysis revealed significant global progress in female education, highlighted by a marked reduction in no-education rates. We observed strong negative correlations between GDP per capita and female no-education rates, indicating that economic development is associated with improved female educational outcomes. However, the correlation between government educational expenditure and GDP growth was weak, suggesting that other variables might be more influential in driving economic growth.

In terms of female workforce participation, we found general growth across most countries, with notable regional variations. Developed nations showed steady, consistent increases, while emerging economies exhibited rapid improvements, reflecting socio-economic and cultural shifts. Our findings underscore the importance of targeted policy efforts, such as paid parental leave, subsidized childcare, and equal pay legislation, in promoting gender equality in the labor market.

Despite these insights, the study acknowledges limitations such as data granularity, the exclusion of unpaid and informal work, and cultural nuances. Future research should address these limitations by incorporating more detailed and intersectional data and considering qualitative methods to better understand cultural influences. Policymakers are encouraged to invest in education, enforce anti-discrimination laws, and promote vocational training to ensure sustained progress toward gender equity and economic growth.

## Data Cleaning: World Bank Education Statistics

- **Loading Data**: Import the EDUCATIONWBSTATS dataset from a CSV file located at `Resources/EDUCATIONWBSTATS.csv`.
- **Handling Missing Values**: Remove rows with missing data to ensure the dataset is complete and reliable.
- **Duplicate Removal**: Check for and identify duplicate rows, ensuring the dataset's integrity.
- **Column Cleanup**: Remove the `Series Code` column to streamline the dataset by eliminating irrelevant information.
- **Column Renaming**: Simplify column names by retaining only the year numbers (1970-2024), enhancing readability.
- **Verification**: Display updated column names to confirm successful changes.

```python
# Load Dataset #1
file_path = 'Resources/EDUCATIONWBSTATS.csv'
WBEDUDATA = pd.read_csv(file_path)

#Drop missing data
WBEDUDATA = WBEDUDATA.dropna()

# Check and print the number of duplicate rows in the DataFrame
num_duplicates = WBEDUDATA.duplicated().sum()
print(f"Number of duplicates: {num_duplicates}")
```

```python
#Removed irrelavant Data
WBEDUDATA = WBEDUDATA.drop(columns=['Series Code'], errors='ignore')

# Rename columns from 1970 to 2024 to only include the year number
WBEDUDATA.columns = [col.split()[0] if col[:4].isdigit() and 1970 <= int(col[:4]) <= 2024 else col for col in WBEDUDATA.columns]

# Display the new column names to verify the changes
print(list(WBEDUDATA.columns))
```

## Data Transformation: World Bank Education Statistics

- **Defining Relevant Series**: Specify a list of relevant series to filter the dataset, focusing on important education and expenditure indicators.
- **Filtering Data**: Filter the DataFrame to include only the relevant series, ensuring the dataset is streamlined for analysis.
- **Resetting Indices**: Reset the DataFrame indices after filtering to maintain a clean and orderly dataset.
- **Melt Transformation**: Transform the DataFrame from wide format to long format using `pd.melt()`, making it suitable for time-series analysis.
- **Ensuring Numeric Year**: Convert the `Year` column to numeric to facilitate chronological analysis.
- **Handling Non-Numeric Values**: Convert non-numeric entries in the `Value` column to numeric, using coercion to handle errors.
- **Pivot Transformation**: Pivot the DataFrame to have `Series` as columns and `Country Name` and `Year` as indices, enabling detailed analysis of each series over time.
- **Verifying Changes**: Print shapes and types of the transformed DataFrames to verify successful transformations.


```python
# Define relevant series
relevant_series = [
    'Barro-Lee: Percentage of female population age 20-24 with no education',
    'Barro-Lee: Percentage of female population age 25+ with no education',
    'Barro-Lee: Percentage of female population age 30-34 with no education',
    'Barro-Lee: Percentage of female population age 35-39 with no education',
    'Barro-Lee: Percentage of female population age 40-44 with no education',
    'Barro-Lee: Percentage of female population age 50-54 with no education',
    'Completion rate, primary education, female (%)',
    'Adult literacy rate, population 15+ years, urban, both sexes (%)',
    'Adult literacy rate, population 15+ years, urban, female (%)',
    'Adult literacy rate, population 15+ years, urban, male (%)',
    'Barro-Lee: Percentage of population age 25+ with no education',
    'Government expenditure on primary education, US$ (millions)',
    'Government expenditure on secondary education, constant US$ (millions)',
    'Government expenditure on tertiary education as % of GDP (%)',
    'Government expenditure on tertiary education, constant US$ (millions)',
    'Government expenditure on secondary education as % of GDP (%)'
]
```

```python
# Filter the DataFrame for relevant series
WBEDUDATA_filteredData = WBEDUDATA[WBEDUDATA['Series'].isin(relevant_series)]

# Output shape of filtered data
filtered_data_shape = WBEDUDATA_filteredData.shape
print(f"Filtered data shape: {filtered_data_shape}")
```

```python
# Reset the indices
WBEDUDATA_filteredData = WBEDUDATA_filteredData.reset_index(drop=True)

# Print filtered dataset
print(WBEDUDATA_filteredData.head())
```

```python
# Print filtered dataset types
print(WBEDUDATA_filteredData.dtypes)
```

```python
# Melt the DataFrame to long format
WBEDUDATA_long = pd.melt(WBEDUDATA_filteredData, id_vars=['Country Name', 'Series'], var_name='Year', value_name='Value')

# Ensure 'Year' is numeric for analysis 
WBEDUDATA_long['Year'] = pd.to_numeric(WBEDUDATA_long['Year'], errors='coerce')

# Handle non-numeric 'Value' entries by converting to numeric and coercing errors 
WBEDUDATA_long['Value'] = pd.to_numeric(WBEDUDATA_long['Value'], errors='coerce')

# Output shape of long data
long_data_shape = WBEDUDATA_long.shape
print(f"Long data shape: {long_data_shape}")
```

```python
# Pivot the DataFrame to have 'Series' as columns
WBEDUDATA_pivot = WBEDUDATA_long.pivot_table(index=['Country Name', 'Year'], columns='Series', values='Value')

# Reset index to flatten the DataFrame
WBEDUDATA_pivot.reset_index(inplace=True)

# Output shape of filtered and pivoted data
pivoted_data_shape = WBEDUDATA_pivot.shape
print(f"Pivoted data shape: {pivoted_data_shape}")
```

```python
# Print filtered dataset types
print(WBEDUDATA_pivot.dtypes)
```

```python
# Display the pivoted DataFrame
WBEDUDATA_pivot.head()
```

### Limitations of the Analysis

Despite the strengths of this study, there are several limitations that should be acknowledged:
1. **Data Granularity**: The data used in this analysis are aggregated at a regional or national level, which may fail to capture intra-region disparities and local variations in female participation rates.
2. **Unpaid and Informal Work**: The study does not account for unpaid labor, such as caregiving and household responsibilities, which disproportionately affect women. Similarly, informal employment—a significant sector for women in some economies—is not fully represented in the dataset.
3. **Cultural Nuances**: Cultural and societal norms, while acknowledged, are difficult to quantify and integrate systematically into the analysis. This introduces potential biases when comparing regions with vastly different societal structures.
4. **Time Dependency**: Workforce participation is influenced by time-sensitive factors, such as temporary economic downturns or policy changes, which are not explicitly analyzed in this study.
5. **Intersectionality**: Factors such as ethnicity, socioeconomic status, and geographic location (urban vs. rural) are not explored in depth but are crucial for a holistic understanding.
